```python
# True duplicates - all fields identical
true_dupes = online_retail_store_df[
    online_retail_store_df.duplicated(
        subset=['Invoice', 'StockCode', 'InvoiceDate', 'Quantity'], 
        keep=False
    )
]
print(true_dupes.shape)

# Conflicting records - same key, different quantity
conflicting = online_retail_store_df[
    online_retail_store_df.duplicated(
        subset=['Invoice', 'StockCode', 'InvoiceDate'], 
        keep=False
    )
] 
conflicting_diff_qty = conflicting[
    conflicting.duplicated(subset=['Invoice', 'StockCode', 'InvoiceDate', 'Quantity'], keep=False) == False
]
print(conflicting_diff_qty.shape)
```
